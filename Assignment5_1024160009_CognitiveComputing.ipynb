```python
from types import MethodDescriptorType
import numpy as np
x = np.array([1,2,3,4,5,1,2,1,1,1])
values, counts = np.unique(x, return_counts=True)
most_frequent = values[np.argmax(counts)]
print("Most frequent value: ", most_frequent)
print("Indices: ", np.where(x == most_frequent)[0])
```

Q4

```python
import numpy as np
Aryan = np.linspace(10, 100, 25)

print("Array: ")
print(Aryan)

print("Dimensions: ", Aryan.ndim)
print("Shape: ", Aryan.shape)
print("Total elements: ", Aryan.size)
print("Data type: ", Aryan.dtype)
print("Total bytes: ", Aryan.nbytes)

reshaped_Aryan = Aryan.reshape(5, 5)
print("Transpose using reshape: ")
print(reshaped_Aryan.T)
```

```python
import numpy as np
ucs420_Aryan = np.array([[10, 20, 30, 40],
                         [50, 60, 70, 80],
                         [90, 15, 20, 35]])
print("Original Array: ")
print(ucs420_Aryan)

print("Mean: ", np.mean(ucs420_Aryan))
print("Median: ", np.median(ucs420_Aryan))
print("Maximum: ", np.max(ucs420_Aryan))
print("Minimum: ", np.min(ucs420_Aryan))
print("Unique Elements: ", np.unique(ucs420_Aryan))
reshaped_ucs420_Aryan = ucs420_Aryan.reshape(4,3)
print("Reshaped array:")
print(reshaped_ucs420_Aryan)

resized_ucs420_Aryan = np.resize(ucs420_Aryan, (2, 3))

print("Resized array:")
print(resized_ucs420_Aryan)
```
