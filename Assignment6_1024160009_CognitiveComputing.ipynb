```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected_temperature = np.add(temperature, 2)
print("Corrected temperatures:", corrected_temperature)

fahrenheit = np.add(np.multiply(corrected_temperature, 9 / 5), 32)
print("Temperatures in Fahrenheit:", fahrenheit)

readings = corrected_temperature[corrected_temperature > 32]
print("Readings greater than 32°C:", readings)

count = np.sum(corrected_temperature > 32)
print("Number of readings exceeding 32°C:", count)

# Vectorization: Performs operations on the entire NumPy array at once, avoiding explicit loops.
# Boolean indexing: Creates a Boolean condition for each element and selects only those satisfying the condition.
```

Q2

```python
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("Total steps:", np.sum(steps))
print("Mean steps:", np.mean(steps))
print("Maximum steps:", np.max(steps))
print("Minimum steps:", np.min(steps))
print("Total steps for each day:", np.sum(steps, axis=0))
print("Total steps for each user:", np.sum(steps, axis=1))
position = np.unravel_index(np.argmax(steps), steps.shape)
print("Position of maximum steps:", position)
```

```python
import numpy as np

original = np.array([1, 2, 3, 4, 5, 6])
subset = original[1:5]
subset[0] = 999

print("Original after modifying subset:", original)
print("Subset:", subset)

copied = original[1:5].copy()
copied[0] = 500
print("Original after modifying copy:", original)
print("Copied array:", copied)

matrix = np.arange(1, 13).reshape(3, 4)
print("3 x 4 matrix:")
print(matrix)

print("First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3:", matrix[0:2, 1:3])

flat_array = matrix.flatten()
ravel_array = matrix.ravel()
print("Flatten array:", flat_array)
print("Ravel array:", ravel_array)

ravel_array[0] = 1000
print("Original after modifying ravel:", matrix)
print("Ravel after modification:", ravel_array)

flat_array[1] = 2000
print("Original after modifying flatten:", matrix)
print("Flatten after modification:", flat_array)

print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

```python
import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)

print("Transpose of X:")
print(X.T)

XtX = X.T @ X
print("X transpose multiplied by X:")
print(XtX)

inverse = np.linalg.inv(XtX)
print("Inverse of X transpose multiplied by X:")
print(inverse)

beta = inverse @ X.T @ y
print("OLS coefficients:", beta)

print("Sleep coefficient:", beta[0])
print("Activity coefficient:", beta[1])
print("Stress coefficient:", beta[2])

"""
Sleep hours: The coefficient is approximately -0.1592.
In this fitted model, increasing sleep by one hour, while keeping the other features constant,is associated with a small decrease in the predicted assistance score.
Activity level: The coefficient is approximately 0.1408.
Increasing activity by one unit, with the other features fixed, increases the predicted score by approximately 0.1408.
Stress level: The coefficient is approximately 10.0630.
Increasing stress by one unit, with the other features fixed, increases the predicted score by approximately 10.063.
"""

new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print("Predicted assistance score:", predicted_score)
```
