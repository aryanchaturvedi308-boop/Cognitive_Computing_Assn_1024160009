```python
import random

roll_no = 1024160009
random.seed(roll_no)

nums = []
for i in range(100):
  nums.append(random.randint(100,900))

count = 0
for i in nums:
  if i % 2 == 1:
    count = count + 1
    print(i)
print(count," odd numbers in the list")

ecount = 0
for i in nums:
  if i % 2 == 0:
    ecount = ecount + 1
    print(i)
print(ecount," even numbers in the list")

def is_prime(n):
  if n < 2:
    return False
  else:
    for i  in range(2, int(n**0.5)+1):
      if n % i == 0:
        return False
  return True

prime_list = []
prime_count = 0
for n in nums:
  if is_prime(n):
    prime_list.append(n)
    prime_count += 1
print(prime_list)
print(prime_count," prime numbers in the list.")

counts = {}
for n in nums:
  if n in counts:
    counts[n] += 1
  else:
    counts[n] = 1
highest_count = 0
most_common = None
for number in counts:
  if counts[number] > highest_count:
    highest_count = counts[number]
    most_common = number
print(highest_count)
print(most_common)
```
